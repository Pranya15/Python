```python
list1 = [1, 2, 3, 4, 5]
list2 = [6, 7, 8, 9, 10]
arr_list2 = np.array([list1, list2])
print(arr_list2)
```

```python
## checking the dimension of the array
print(arr_list2.ndim)
```

```python
# numpy array attributes
arr1 = np.array([[1,2,3],[4,5,6]])
print(arr1)
```

```python
print("Shape of the array:", arr1.shape)
print("Size of the array:", arr1.size)
print("Data type of the array:", arr1.dtype)
print("Number of dimensions of the array:", arr1.ndim)
```
